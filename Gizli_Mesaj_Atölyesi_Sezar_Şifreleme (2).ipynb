## 3. Sezar Şifreleme Kuralımız

Bu uygulamada 29 harfli Türk alfabesini kullanacağız:

A B C Ç D E F G Ğ H I İ J K L M N O Ö P R S Ş T U Ü V Y Z

- Mesajlarımızı büyük harflerle yazacağız.
- Anahtar, her harfi kaç adım ileri kaydıracağımızı belirleyecek.
- Alfabenin sonuna gelirsek başından devam edeceğiz.
- Boşlukları ve noktalama işaretlerini değiştirmeyeceğiz.
- Mesajı çözmek için anahtar kadar geri gideceğiz.

**Örnek: Anahtar = 3**

A → B (1) → C (2) → Ç (3)

A harfinin şifrelenmiş karşılığı Ç olur.
Başladığımız harfi adım olarak saymayız.

## 4. Algoritma ile Anahtarın Farkı

**Algoritma:** Her harfi alfabede belirli sayıda ileri kaydır.

ancak ortaya çıkan şifreli mesaj değişir.

## 2. Harfleri Sayılarla Temsil Edelim

Python’da bir metindeki her karakterin konum numarası vardır.
Bu numaraya **indeks** denir. İndeksler 0’dan başlar.

Örneğin Türk alfabesinde:

| Harf | A | B | C | Ç | D | E |
|---|---|---|---|---|---|---|
| İndeks | 0 | 1 | 2 | 3 | 4 | 5 |

Bir harfi şifrelemek için:

1. Harfin alfabedeki indeksini bul.
2. İndekse anahtarı ekle.
3. Bulduğun yeni konumdaki harfi oku.

**Tahmin:** A harfinin indeksi 0 ise anahtar 3 olduğunda
hangi indekse ve hangi harfe ulaşırız?

```python
alfabe = "ABCÇDEFGĞHIİJKLMNOÖPRSŞTUÜVYZ"

harf = "A"
anahtar = 3

konum = alfabe.index(+,f)
yeni_konum = konum + anahtar
sifreli_harf = alfabe[yeni_konum]

print("Harf:", harf)
print("Eski indeks:", konum)
print("Yeni indeks:", yeni_konum)
print("Şifreli harf:", sifreli_harf)
```

DİKKAT:

| Kod                  | Görevi                             |
| -------------------- | ---------------------------------- |
| `alfabe.index(harf)` | Harfin alfabedeki indeksini bulur. |
| `konum + anahtar`    | Anahtar kadar ileri gider.         |
| `alfabe[yeni_konum]` | Yeni indeksteki harfi getirir.     |


ÖNEMLİ:

Z harfinin indeksi 28. Üç ekleyince 31 oluyor. Ancak alfabemizdeki son indeks 28. Alfabenin başına nasıl dönebiliriz?

### Alfabenin Sonundan Başına Dönmek

Türk alfabesinde 29 harf vardır. Geçerli indeksler 0–28 arasındadır.

Alfabenin sonunu geçtiğimizde başına dönmek için
**kalan bulma operatörünü (%)** kullanabiliriz.

31 % 29 = 2

Çünkü 31’in 29’a bölümünden kalan 2’dir.
Alfabemizde 2 indeksinde C vardır.

Bu nedenle anahtar 3 olduğunda:

Z → A (1) → B (2) → C (3)

Yeni indeks hesabımız:

yeni_konum = (konum + anahtar) % len(alfabe)

len(alfabe), alfabedeki karakter sayısını verir: 29.
Buradaki % işareti yüzde hesaplamaz, bölümden kalanı bulur.

```python
harf = "Z"
anahtar = 3

konum = alfabe.index(harf)
yeni_konum = (konum + anahtar) % len(alfabe)
sifreli_harf = alfabe[yeni_konum]

print("Harf:", harf)
print("Kaydırma sonrası değer:", konum + anahtar)
print("Alfabeye döndürülmüş indeks:", yeni_konum)
print("Şifreli harf:", sifreli_harf)
```

DENEME:

 Aşağıdaki durumlarda sonucu önce tahmin edelim, sonra kodla kontrol edelim. Ö, A, A.

| Harf | Anahtar |
| ---- | ------: |
| M    |       3 |
| Y    |       2 |
| A    |      29 |


| Adım         | Harf | İndeks |
| ------------ | ---- | -----: |
| Başlangıç    | Z    |     28 |
| 1 adım ileri | A    |      0 |
| 2 adım ileri | B    |      1 |
| 3 adım ileri | C    |      2 |


yeni_konum = (28 + 3) % 29
# 31 % 29 = 2 → C

3. Aşama — Mesajın tamamını şifreleyelim

Tek bir harfi şifreleyebiliyoruz. Şimdi aynı işlemi mesajdaki her harfe sırayla uygulayacağız. Bunun için for döngüsünü kullanacağız.

## 3. Mesajın Tamamını Şifreleyelim

Görevimiz: MERHABA mesajını, anahtar 3 olacak şekilde şifrelemek.

Algoritmamız:

1. Alfabeyi, mesajı ve anahtarı belirle.
2. Şifreli mesajı biriktirmek için boş bir metin oluştur.
3. Mesajdaki karakterleri sırayla incele:
   - Karakter alfabede varsa anahtar kadar ileri kaydır.
   - Alfabede yoksa değiştirmeden bırak.
4. Elde edilen karakteri şifreli mesajın sonuna ekle.
5. Döngü bitince şifreli mesajı göster.

Bu uygulamada mesajı Türk alfabesindeki BÜYÜK HARFLERLE yazacağız.
Boşlukları, rakamları ve noktalama işaretlerini değiştirmeyeceğiz.

```python
alfabe = "ABCÇDEFGĞHIİJKLMNOÖPRSŞTUÜVYZ"
mesaj = "MERHABA"
anahtar = 3

sifreli_mesaj = ""

for harf in mesaj:
    if harf in alfabe:
        konum = alfabe.index(harf)
        yeni_konum = (konum + anahtar) % len(alfabe)
        yeni_harf = alfabe[yeni_konum]
    else:
        yeni_harf = harf

    sifreli_mesaj += yeni_harf
    print(harf, "→", yeni_harf)

print("Şifreli mesaj:", sifreli_mesaj)
```

| Kod                          | Açıklama                                                                           |
| ---------------------------- | ---------------------------------------------------------------------------------- |
| `sifreli_mesaj = ""`         | Başlangıçta boş bir metin oluşturur. Dönüştürdüğümüz harfleri burada biriktiririz. |
| `for harf in mesaj:`         | Mesajdaki karakterleri sırayla ele alır.                                           |
| `if harf in alfabe:`         | Karakterin tanımladığımız alfabede bulunup bulunmadığını kontrol eder.             |
| `else:`                      | Boşluk gibi alfabede bulunmayan karakterleri değiştirmeden bırakır.                |
| `sifreli_mesaj += yeni_harf` | Yeni karakteri, biriken şifreli mesajın sonuna ekler.                              |


| İşlenen harf | Eski indeks | Anahtar eklenince | Yeni indeks (`% 29`) | Yeni harf | Biriken mesaj |

| M            |          15 |                18 |                   18 | Ö         | Ö             |

4. Aşama — Şifreli mesajı çözelim

Şifrelerken harfleri anahtar kadar ileri kaydırdık. Şimdi aynı anahtarla geri kaydırarak asıl mesajı bulacağız.

## 4. Şifreli Mesajı Çözelim

Elimizdeki şifreli mesaj: ÖĞTJÇDÇ
Anahtarımız: 3

Şifrelerken harfin indeksine anahtarı ekledik.
Çözerken şifreli harfin indeksinden anahtarı çıkaracağız.

Şifreleme:
yeni_konum = (konum + anahtar) % len(alfabe)

Şifre çözme:
yeni_konum = (konum - anahtar) % len(alfabe)

Örnek:
Ö harfinin indeksi 18’dir.
18 - 3 = 15
15 indeksindeki harf M’dir.

Her harfe aynı işlemi uygulayarak mesajın tamamını çözeceğiz.

Çözülen mesaj: MERHABA

```python
alfabe = "ABCÇDEFGĞHIİJKLMNOÖPRSŞTUÜVYZ"
sifreli_mesaj = "ÖĞTJÇDÇ"
anahtar = 3
cozulen_mesaj = ""

for harf in sifreli_mesaj:
    if harf in alfabe:
        konum = alfabe.index(harf)
        yeni_konum = (konum - anahtar) % len(alfabe)
        cozulen_mesaj += alfabe[yeni_konum]
    else:
        cozulen_mesaj += harf

print("Çözülen mesaj:", cozulen_mesaj)
```

| Şifreli harf | Eski indeks | Anahtar çıkarılınca | Yeni indeks (`% 29`) | Çözülen harf |
| ------------ | ----------: | ------------------: | -------------------: | ------------ |
| Ö            |          18 |                  15 |                   15 | M            |
| Ğ            |           8 |                   5 |                    5 | E            |
| T            |          23 |                  20 |                   20 | R            |
| J            |          12 |                   9 |                    9 | H            |
| Ç            |           3 |                   0 |                    0 | A            |
| D            |           4 |                   1 |                    1 | B            |
| Ç            |           3 |                   0 |                    0 | A            |

## 5. Anahtarı Bulalım

Bir şifreli mesaj ele geçirdik: ÖĞTJÇDÇ

Kullanılan Türk alfabesini ve Sezar şifreleme yöntemini biliyoruz.
Ancak anahtarı bilmiyoruz.

Türk alfabesinde 29 harf olduğu için 0–28 arasındaki bütün
kaydırmaları deneyebiliriz. Daha büyük anahtarlar aynı
kaydırmaları tekrar eder.

Olası bütün anahtarları deneme yaklaşımına
**kaba kuvvet (brute force)** denir.

Görevimiz: Her anahtarla elde edilen metni inceleyerek
hangisinin anlamlı bir mesaj olduğunu bulmak.

Range kullanımı:

range(3) → “0’dan başla, 3’e gelmeden dur.”

Bu yüzden sayi ilk turda 0, ikinci turda 1, üçüncü turda 2 olur. 3 dahil değildir.

```python
for sayi in range(3):
    print(sayi)
```

```python
alfabe = "ABCÇDEFGĞHIİJKLMNOÖPRSŞTUÜVYZ"
sifreli_mesaj = "ÖĞTJÇDÇ"

for anahtar in range(len(alfabe)):
    cozulen_mesaj = ""

    for harf in sifreli_mesaj:
        if harf in alfabe:
            konum = alfabe.index(harf)
            yeni_konum = (konum - anahtar) % len(alfabe)
            cozulen_mesaj += alfabe[yeni_konum]
        else:
            cozulen_mesaj += harf

    print(anahtar, "→", cozulen_mesaj)
```

| Kod                                  | Görevi                                                    |
| ------------------------------------ | --------------------------------------------------------- |
| `for anahtar in range(len(alfabe)):` | 0’dan 28’e kadar bütün anahtarları dener.                 |
| `cozulen_mesaj = ""`                 | Her yeni anahtarda biriken mesajı sıfırlar.               |
| `for harf in sifreli_mesaj:`         | O anahtarı kullanarak mesajdaki bütün karakterleri işler. |

Her turda mesajın tamamı o anahtarla çözülmeye çalışılır. Böylece 29 aday mesaj elde edilir.
